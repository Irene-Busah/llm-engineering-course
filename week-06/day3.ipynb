## Building Models with Traditional ML & XGBoost

#### Evaluation, Baselines & Traditional ML

```python
# libraries
import random
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.ensemble import RandomForestRegressor
from pricer.evaluator import evaluate
from pricer.items import Item
```

```python
LITE_MODE = False
```

```python
# loading the dataset
username = 'ed-donner'
dataset = f"{username}/items_lite" if LITE_MODE else f"{username}/items_full"

# retrieving the train, test and validation data
train, val, test = Item.from_hub(dataset)

print(f"Loaded {len(train):,} training items, {len(val):,} validation items, {len(test):,}")
```

```python
def random_pricer(item):
    return random.randrange(1, 1000)
```

```python
random.seed(42)
evaluate(random_pricer, test)
```

```python
training_prices = [item.price for item in train]
training_average = sum(training_prices) / len(training_prices)
print(training_average)

def constant_pricer(item):
    return training_average
```

```python
evaluate(constant_pricer, test)
```

```python
def get_features(item):
    return {
        "weight": item.weight,
        "weight_unknown": 1 if item.weight==0 else 0,
        "text_length": len(item.summary)
    }
```

```python
# print(train[0])
```

```python
def list_to_dataframe(items):
    features = [get_features(item) for item in items]

    dataframe = pd.DataFrame(features)
    dataframe['price'] = [item.price for item in items]

    return dataframe


train_df = list_to_dataframe(train)
test_df = list_to_dataframe(test)
```

```python
# building the traditional linear regression
np.random.seed(42)

# separating features and target
feature_columns = ['weight', 'weight_unknown', 'text_length']

x_train = train_df[feature_columns]
y_train = train_df['price']

x_test = test_df[feature_columns]
y_test = test_df['price']


# training the linear model
model = LinearRegression()
model.fit(x_train, y_train)


for feature, coef in zip(feature_columns, model.coef_):
    print(f"{feature}: {coef}")

print(f"Intercept: {model.intercept_}")

# predict on the test set and evaluate
y_pred = model.predict(x_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse}")
print(f"R-Squared Score: {r2}")
```

```python
def linear_regression_pricer(item):
    features = get_features(item)

    features_df = pd.DataFrame([features])

    return model.predict(features_df)[0]
```

```python
evaluate(linear_regression_pricer, test)
```

```python
prices = np.array([float(item.price) for item in train])
documents = [item.summary for item in train]
```

```python
np.random.seed(42)
vectorizer = CountVectorizer(max_features=2000, stop_words='english')
x = vectorizer.fit_transform(documents)
```

```python
selected_words = vectorizer.get_feature_names_out()
print(f"Number of Selected words: {len(selected_words)}")
print(f"Selected Words: {selected_words[1000:1020]}")
```

```python
regressor = LinearRegression()
regressor.fit(x, prices)
```

```python
def natural_language_linear_regression_pricer(item):
    x = vectorizer.transform([item.summary])

    return max(regressor.predict(x)[0], 0)
```

```python
evaluate(natural_language_linear_regression_pricer, test)
```

```python
subset = 15_000
random_forest_model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=4)
random_forest_model.fit(x[:subset], prices[:subset])
```

```python
def random_forest(item):
    x = vectorizer.transform([item.summary])
    return max(0, random_forest_model.predict(x)[0])
```

```python
evaluate(random_forest, test)
```

### Using XGBoost

```python
import xgboost as xgb
```

```python
np.random.seed(42)

xgb_model = xgb.XGBRegressor(n_estimators=1000, random_state=42, n_jobs=4, learning_rate=0.1)
xgb_model.fit(x, prices)
```

```python
def xg_boost(item):
    x = vectorizer.transform([item.summary])

    return max(0, xgb_model.predict(x)[0])
```

```python
evaluate(xg_boost, test)
```
