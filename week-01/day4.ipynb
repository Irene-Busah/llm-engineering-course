### Tokenizing with code

```python
import tiktoken

encoding = tiktoken.encoding_for_model('gpt-4.1-mini')

tokens = encoding.encode("Hi, my name is Irene")

tokens
```

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")
```

## Illusion of Memory

```python
from openai import OpenAI

openai = OpenAI()
```

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi, I am Irene!"}
]

response = openai.chat.completions.create(model='gpt-4.1-mini', messages=messages)
response.choices[0].message.content
```

```python
followup = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "What is my name?"}
]

response = openai.chat.completions.create(model='gpt-4.1-mini', messages=followup)
response.choices[0].message.content
```

```python
followup = [
    {"role": "system", "content": "You are a helpful assistant"},
    {"role": "user", "content": "Hi, I am Irene!"},
    {"role": "assistant", "content": "Hello Irene! How can I assist you today?"},
    {"role": "user", "content": "What is my name?"},
]

response = openai.chat.completions.create(model='gpt-4.1-mini', messages=followup)
response.choices[0].message.content
```
